## TASK 1 - Explore the menu

```python
# 1 Print the full menu grouped by category,

# First collect all unique categories from the menu
categories = set()  #Uses a set to collect unique categories

for item in menu:
    categories.add(menu[item]["category"])

# Loop through each category
for category in categories:

    print("=====", category, "=====")

    # Loop through menu items and print items belonging to that category
    for item, details in menu.items():

        if details["category"] == category:

            price = details["price"]

            if details["available"]:
                status = "Available"
            else:
                status = "Unavailable"

            print(f"{item:<15} ₹{price:.2f}   [{status}]")

    print()
```

```python
# 2 Using dictionary methods, compute:

#  Get the total number of items
len_menu = len(menu)
print(f"Total number of items on the menu: {len_menu}")

# Total number of available items

available_count =0
for item in menu:
    if menu[item]["available"] == True:
        available_count +=1
print(f"Total number of available items: {available_count}")

# The most expensive item (name + price)
most_expensive = max(menu.items(), key=lambda x: x[1]["price"])

print("Most expensive item:", most_expensive[0], f"(₹{most_expensive[1]['price']:.2f})")

# All items priced under ₹150 (name + price)
print("\nItems priced under ₹150:")
for item, details in menu.items():
    if details["price"]<150:
        print(f"{item:<20}- ₹{details['price']:.2f}")

```

## Task 2 — Cart Operations

```python
cart=[]
# 1 Write logic to add an item to the cart:
def add_to_cart(item_name,quantity):
  # Check if the item exists
    if item_name not in menu:
        print(f"{item_name} not found in menu")
        return

  # Check availability
    if not menu[item_name]["available"]:
        print(f"{item_name} is currently unavailable")
        return

  # Check if already in the cart
    for item in cart:
        if item["item"]==item_name:
           item["quantity"]+=quantity
           print(f"Updated {item_name} quantity to {item["quantity"]}")
           return

   # Add new item
    cart.append({"item":item_name,
                "quantity":quantity,
                "price":menu[item_name]["price"]})

    print(f"Added {item_name} x {quantity}")


# Function to remove the item
def remove_from_cart(item_name):
    for item in cart:
        if item["item"]==item_name:
            cart.remove(item)
            print(f"Removed {item_name}")
            return
    print(f"{item_name} not found in cart")


# Function to update quantity
def update_quantity(item_name,quantity):
    for item in cart:
      if item["item"]== item_name:
         item["quantity"]= quantity
         print(f"{item_name} quantity updated to{quantity}")
         return
    print(f"{item_name} not found in cart")

# Function to print cart
def show_cart():
    print("\nCurrent Cart:")
    for item in cart:
        print(item)
    print("-" * 30)


add_to_cart("Paneer Tikka", 2)
show_cart()

add_to_cart("Gulab Jamun", 1)
show_cart()

add_to_cart("Paneer Tikka", 1)   # should update to 3
show_cart()

add_to_cart("Mystery Burger", 1)  # not in menu
show_cart()

add_to_cart("Chicken Wings", 1)   # unavailable
show_cart()

remove_from_cart("Gulab Jamun")
show_cart()

print("\n =========Order Summary =========")

subtotal=0
for item in cart:
    total =item["price"]*item["quantity"]
    subtotal+=total
    print(f"{item['item']:15} x {item['quantity']} {total:.2f}")

gst = subtotal * 0.05
total_payable = subtotal + gst

print("------------------------------------")
print(f"Subtotal: ₹{subtotal:.2f}")
print(f"GST (5%): ₹{gst:.2f}")
print(f"Total Payable: ₹{total_payable:.2f}")
print("====================================")

```


## Task 3 Inventory tracker with deep copy

```python
import copy
#1.Deep copy
inventory_backup = copy.deepcopy(inventory)
# Change one value to test deep copy
inventory["Paneer Tikka"]["stock"]=5

print("Current inventory:",inventory["Paneer Tikka"])
print("Backup inventory:",inventory_backup["Paneer Tikka"])

#Restore Inventory
inventory = copy.deepcopy(inventory_backup)
```

```python
# 2.Deduct quantities based on cart
for item in cart:
    name= item["item"]
    qty=item["quantity"]

    available_stock= inventory[name]["stock"]

    if available_stock >= qty:
        inventory[name]["stock"]-=qty
        print(f"{name}:stock reduced by {qty}")
    else:
        print(f"not enough stock for {name}.Only {available_stock} left.")
        inventory[name]["stock"]= 0

# 3 Reorder alerts
print("\nReorder Alerts:")

for item, details in inventory.items():
    if details["stock"]<= details["reorder_level"]:
        print(f"⚠ Reorder Alert:{item}- only {details['stock']} unit(s) left (reorder level:{details["reorder_level"]})")

# 4. Show final comparison
print("\n Final Inventory vs Backup:")

print("\nCurrent Inventory:")
for item, details in inventory.items():
    print(item, details)

print("\nBackup Inventory (unchanged):")
for item, details in inventory_backup.items():
    print(item, details)
```

## Task 4 — Daily Sales Log Analysis

```python
# 1 Revenue per day

# ---------------- SALES LOG ANALYSIS ----------------

# 1. Revenue per day
print("\nRevenue per Day:")

daily_revenue = {}

for date, orders in sales_log.items():
    total = 0
    for order in orders:
        total += order["total"]

    daily_revenue[date] = total
    print(f"{date} : ₹{total}")


# 2. Best selling day
best_day = max(daily_revenue, key=daily_revenue.get)
print("\nBest Selling Day:", best_day, "-", daily_revenue[best_day])


# 3. Most ordered item
item_count = {}

for orders in sales_log.values():
    for order in orders:
        for item in order["items"]:
            if item in item_count:
                item_count[item] += 1
            else:
                item_count[item] = 1

most_ordered = max(item_count, key=item_count.get)
print("Most Ordered Item:", most_ordered)


# 4. Add new day
sales_log["2025-01-05"] = [
    {"order_id": 11, "items": ["Butter Chicken", "Gulab Jamun", "Garlic Naan"], "total": 490.0},
    {"order_id": 12, "items": ["Paneer Tikka", "Rasgulla"], "total": 260.0}
]

print("\nAfter adding new day:")

# Recalculate revenue
daily_revenue = {}

for date, orders in sales_log.items():
    total = sum(order["total"] for order in orders)
    daily_revenue[date] = total
    print(f"{date} : ₹{total}")

# New best day
best_day = max(daily_revenue, key=daily_revenue.get)
print("\nNew Best Selling Day:", best_day, "-", daily_revenue[best_day])

# 5. Numbered list of all orders
print("\nAll Orders:")
all_orders = []

# Flatten all orders into one list
for date, orders in sales_log.items():
    for order in orders:
        all_orders.append((date, order))

# Use enumerate
for i, (date, order) in enumerate(all_orders, start=1):
    items = ", ".join(order["items"])
    print(f"{i}. [{date}] Order #{order['order_id']} — ₹{order['total']} — Items: {items}")


```
